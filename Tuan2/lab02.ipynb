# Autograd cơ bản

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

### Tính thủ công đạo hàm của y = $x^2$ tại x = 3.6 và so sánh với x.grad

- y = f(x) = $x^2$
- Đạo hàm bậc nhất của hàm số:  y' = $\frac{dx}{dy}$ = 2x
- Thay x = 3.6: y'(3.6) = 2 x 3.6 = 7.2
- Vậy hai kết quả giống nhau

### Đổi hàm thành y = 3*$x^2$ + 2*x + 1 và kiểm tra gradient

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = 3*x**2 + 2*x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

# Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

### So sánh ba hệ số học được với [1, 2, 3]:

- Hệ số bậc hai: Hội tụ rất nhanh và chuẩn xác về 1
- Hệ số bậc một: Hội tụ tốt về giá trị 2
- Hệ số tự do: Thường hội tụ chậm hơn hai hệ số trên sau 1000 epoch

### Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step()

- optimizer.zero_grad(): xóa gradient cũ về 0 
- mse.backward(): lan truyền ngược - tính đạo hàm riêng của hàm mất mát đối với từng tham số có requires_grad=True
- optimizer.step(): cập nhật trọng số - điều chỉnh w theo hướng làm giảm hàm mất mát MSE nhất có thể cho vòng lặp tiếp theo

# Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

### Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

- Thỏa mãn gần như tuyệt đối (sai số ở bậc $10^-$ đến $10^-5$)

### Nhận xét vì sao bài toán có thể có nhiều nghiệm.

- Hàm mục tiêu $sqerr(A, B, C, D)$ không có một điểm cực tiểu toàn cục duy nhất, mà tạo thành một đường thẳng đáy phẳng (valley) gồm các điểm mà tại đó $sqerr = 0$. Thuật toán NAdam xuất phát từ điểm ngẫu nhiên nào sẽ bị kéo về điểm nghiệm $sqerr = 0$ gần điểm xuất phát đó nhất theo hướng dốc của gradient.

# Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:
+ Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
+ Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
+ Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
+ Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

torch.manual_seed(42)
np.random.seed(42)

print("=" * 60)
print("YÊU CẦU 1: TẠO VÀ RESHAPE TENSOR X")
print("=" * 60)

X = torch.arange(24)  # Tạo tensor gồm 24 số nguyên từ 0 đến 23
print(f"Tensor X ban đầu (shape {X.shape}):\n{X}\n")

X_4x6 = X.reshape(4, 6)
print(f"Reshape (4 x 6) - shape {X_4x6.shape}:\n{X_4x6}\n")

X_2x12 = X.reshape(2, 12)
print(f"Reshape (2 x 12) - shape {X_2x12.shape}:\n{X_2x12}\n")

X_2x3x4 = X.reshape(2, 3, 4)
print(f"Reshape (2 x 3 x 4) - shape {X_2x3x4.shape}:\n{X_2x3x4}\n")

print("=" * 60)
print("YÊU CẦU 2: CÁC PHÉP TOÁN TRÊN TENSOR A, B (3x4)")
print("=" * 60)

A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("Tensor A:\n", A)
print("\nTensor B:\n", B)

print("\n1. Tổng (A + B):\n", A + B)
print("\n2. Tích từng phần tử (A * B):\n", A * B)
print("\n3. Thương từng phần tử (A / B):\n", A / B)

print("\n4. Trung bình theo cột của A (dim=0):\n", torch.mean(A, dim=0))
print("\n5. Độ lệch chuẩn theo cột của A (dim=0):\n", torch.std(A, dim=0))

print("YÊU CẦU 3: TÍNH ĐẠO HÀM BẰNG AUTOGRAD")
print("=" * 60)

x_val = torch.tensor(2.0, requires_grad=True)
y_val = 2 * x_val**2 + 5 * x_val + 3
y_val.backward()

print(f"Giá trị x = {x_val.item()}")
print(f"Giá trị y = {y_val.item()}")
print(f"Đạo hàm dy/dx bằng PyTorch Autograd: {x_val.grad.item()}")

print("\n" + "=" * 60)
print("YÊU CẦU 4: HỌC HỆ SỐ CỦA ĐA THỨC y = 2x^2 - 3x + 5")
print("=" * 60)

polynomial = np.poly1d([2, -3, 5])
N = 20
X_poly = np.random.randn(N, 1) * 5
Y_poly = polynomial(X_poly)

XX = np.hstack([X_poly * X_poly, X_poly, np.ones_like(X_poly)])

x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y_poly, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)
print(f"Hệ số khởi tạo ban đầu:\n{w.detach().numpy().flatten()}\n")

optimizer = torch.optim.NAdam([w], lr=0.05)

for epoch in range(3000):
  optimizer.zero_grad()
  y_pred = x_train @ w
  mse = torch.mean(torch.square(y_train - y_pred))
  mse.backward()
  optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print(f"Loss MSE sau huấn luyện: {mse.item():.6e}")
print(f"Các hệ số tìm được: {learned_coeffs}")
print(
    f"So sánh với hệ số gốc [2, -3, 5]: sai số tối đa <"
    f" {np.max(np.abs(learned_coeffs - np.array([2, -3, 5]))):.6f}"
)
```
