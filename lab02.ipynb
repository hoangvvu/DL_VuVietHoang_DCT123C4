### Tạo Tensor cơ bản

```python
import torch

a = torch.tensor(
    [[1,2,3],
     [4,5,6]],
    dtype=torch.int32
)

print(a)
```

- kích thước: 2x3
- số chiều: 2
- kiểu dữ liệu: torch.int32 (số nguyên 32-bit)

### Tạo các Tensor thường dùng

```python
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)
```

- Sự khác nhau giữa torch.rand() và torch.randn():
| Đặc điểm| torch.rand()|torch.randn()|
|--------:|------------:|------------:|
|**Phân phối**|Phân phối đều|Phân phối chuẩn hóa|
|**Khoảng giá trị**|[0,1)|(-$\infty$,+$\infty$)|
|**Giá trị trung bình**|$\approx$ 0.5|$\approx$ 0|
|**Độ lệch chuẩn**|$\approx$ 0.289|1|
|**Ký tự nhận biết**|rand = random| randn = random normal|

- Tạo thêm một tensor 5x5 chứa toàn giá trị 10

```python
x = torch.full((5,5), 10)
print(x)
```

### Kiểm tra thông tin Tensor

```python
a = torch.zeros(2,3,4)
print("Tensor: ")
print("Shape: ", a.shape)
print("Size: ", a.size())
print("Number of dimensions: ", a.ndim)
print("Data type: ", a.dtype)
print(a)
```

- Ý nghĩa của torch.Size([2, 3, 4]): độ dày = 2, chiều cao = 3, chiều rộng = 4
